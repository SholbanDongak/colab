# Часть 1

Почему учет цензурированных случаев важен?

Распространенная ошибка заключается в том, что при анализе данных игнорируются наблюдения (те элементы выборки), для которых терминальное событие не наступило.

Рассмотрим случай, когда популяция фактически состоит из двух субпопуляций, $A$ и $B$.
* субпопуляция $A$ имеет очень маленькую продолжительность жизни, например, в среднем 2 года,
* субпопуляция $B$ имеет гораздо большую продолжительность жизни, скажем, в среднем 12 лет.

Мы не знаем этого различия заранее. При $t=10$, мы хотим исследовать среднюю продолжительность жизни для всей популяции.

На рисунке красными линиями обозначена продолжительность жизни особей, у которых наблюдалось терминальное событие, а синими линиями обозначена продолжительность жизни особей, для которых терминальное событие не наблюдалось (цензурированных справа).

Если необходимо оценить среднюю продолжительность жизни в данной популяции, и мы наивно решим не включать особей сензурированных справо, то мы, очевидно, недооценим истинную среднюю продолжительность жизни. То есть полученная оценка будет смещенной (*англ*. biased) влево.

```python
from lifelines.plotting import plot_lifetimes
import numpy as np
from numpy.random import uniform, exponential

np.random.seed(1)
N = 25

CURRENT_TIME = 10

actual_lifetimes = np.array([
    exponential(12) if (uniform() < 0.5) else exponential(2) for i in range(N)
])
observed_lifetimes = np.minimum(actual_lifetimes, CURRENT_TIME)
death_observed = actual_lifetimes < CURRENT_TIME

ax = plot_lifetimes(observed_lifetimes, event_observed=death_observed)

ax.set_xlim(0, 25)
ax.vlines(10, 0, 30, lw=2, linestyles='--')
ax.set_xlabel("time")
ax.set_title("Births and deaths of our population, at $t=10$")
print("Observed lifetimes at time %d:\n" % (CURRENT_TIME), observed_lifetimes)
```

$$
H(t) = \int_0^t h(\tau) d \tau
$$

```python
T = data["duration"]
E = data["observed"]

from lifelines import NelsonAalenFitter
naf = NelsonAalenFitter()

naf.fit(T,event_observed=E)
```

```python
print(naf.cumulative_hazard_.head())
naf.plot_cumulative_hazard()
```

```python
dem = (data["democracy"] == "Democracy")
naf.fit(T[dem], event_observed=E[dem], label="Democratic Regimes")
ax = naf.plot_cumulative_hazard(loc=slice(0, 20))

naf.fit(T[~dem], event_observed=E[~dem], label="Non-democratic Regimes")
naf.plot_cumulative_hazard(ax=ax, loc=slice(0, 20))

plt.title("Cumulative hazard function of different global regimes");
```

```python
bandwidth = 3.

naf.fit(T[dem], event_observed=E[dem], label="Democratic Regimes")
ax = naf.plot_hazard(bandwidth=bandwidth)

naf.fit(T[~dem], event_observed=E[~dem], label="Non-democratic Regimes")
naf.plot_hazard(ax=ax, bandwidth=bandwidth)

plt.title("Hazard function of different global regimes | bandwidth=%.1f" % bandwidth);
plt.ylim(0, 0.4)
plt.xlim(0, 25);
```

# Оценка кумулятивных рисков с использованием параметрических моделей

Сценарий: 500 пациентов с ишемической болезнью сердца. Им установлены стенты с лекарственным покрытием двух типов:

Старое покрытие (Control): Элюирование препарата прекращается через 6 месяцев, риск рестеноза начинает расти.

Новое покрытие (Drug): Пролонгированное высвобождение, риск рестеноза снижен и медленнее растёт во времени.

Событие: Рестеноз (повторное сужение сосуда), выявленный при контрольной ангиографии.
Цензурирование: Пациенты выбывают из наблюдения или исследование заканчивается.

Лучше всего иллюстрировать кумулятивный риск
$$
H(t) = ( t / \lambda)^ \rho,
$$
потому что в логарифмических координатах это дает идеальную прямую — наглядная верификация модели.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from lifelines import WeibullFitter
from lifelines.plotting import qq_plot

# Для воспроизводимости
np.random.seed(2026)

# Параметры Вейбулла: масштаб lambda и форма k
# Контроль: базовый риск, форма = 1.5 (риск растет)
lambda_control = 18.0  # месяцев
k_control = 1.5

# Новое покрытие: увеличиваем масштаб (медленнее) + чуть меняем форму
lambda_drug = 24.0
k_drug = 1.3

# Генерируем времена событий для 250 пациентов в каждой группе
t_control = np.random.weibull(k_control, 250) * lambda_control
t_drug = np.random.weibull(k_drug, 250) * lambda_drug

# Цензурирование: административное (исследование до 36 мес) + потери
censor_time = np.random.uniform(6, 40, 500)
censor_control = t_control > censor_time[:250]
censor_drug = t_drug > censor_time[250:]

# Итоговое время наблюдения
T_control = np.minimum(t_control, censor_time[:250])
T_drug = np.minimum(t_drug, censor_time[250:])
E_control = ~censor_control  # 1 = событие, 0 = цензура
E_drug = ~censor_drug

# Собираем датасет
df_control = pd.DataFrame({
    'time': T_control,
    'event': E_control.astype(int),
    'group': 'Control'
})
df_drug = pd.DataFrame({
    'time': T_drug,
    'event': E_drug.astype(int),
    'group': 'Drug'
})
df = pd.concat([df_control, df_drug], ignore_index=True)

print(df.head())
print(f"Доля событий: {df['event'].mean():.2f}")
```

```python
# Раздельный анализ для двух групп
wb_control = WeibullFitter().fit(df_control['time'],
                                 df_control['event'],
                                 label='Старый стент')
wb_drug = WeibullFitter().fit(df_drug['time'],
                             df_drug['event'],
                             label='Новый стент')

# Вывод параметров
print("Контрольная группа:")
print(f"  lambda (масштаб): {wb_control.lambda_:.2f} мес")
print(f"  rho (форма): {wb_control.rho_:.2f}")
# Медиана для распределения Вейбулла: λ * (ln 2)^(1/ρ)
median_control = wb_control.lambda_ * np.log(2)**(1/wb_control.rho_)
print(f"  Медиана выживаемости: {median_control:.1f} мес")

print("\nГруппа с новым покрытием:")
print(f"  lambda: {wb_drug.lambda_:.2f} мес")
print(f"  rho: {wb_drug.rho_:.2f}")
median_drug = wb_drug.lambda_ * np.log(2)**(1/wb_drug.rho_)
print(f"  Медиана выживаемости: {median_drug:.1f} мес")

# Визуализация 1: Кумулятивный риск H(t) = (t/λ)^ρ
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Левый график: сами функции кумулятивного риска
ax = axes[0]
wb_control.cumulative_hazard_.plot(ax=ax, drawstyle='steps-post', lw=2)
wb_drug.cumulative_hazard_.plot(ax=ax, drawstyle='steps-post', lw=2)
ax.set_title('Кумулятивный риск рестеноза', fontsize=12)
ax.set_xlabel('Время (месяцы)')
ax.set_ylabel('H(t)')
ax.grid(alpha=0.3)

# Правый график: лог-лог преобразование — должно быть линейным!
ax = axes[1]
for wb, label in [(wb_control, 'Старый стент'), (wb_drug, 'Новый стент')]:
    # Берем наблюдаемые точки, избегаем H=0
    h = wb.cumulative_hazard_.iloc[:, 0]
    t = h.index
    mask = (h > 0) & (t > 0)
    log_t = np.log(t[mask])
    log_h = np.log(h[mask])
    ax.plot(log_t, log_h, 'o', markersize=3, label=label)

    # Теоретическая прямая: ln H = k * ln t - k * ln λ
    t_theory = np.linspace(1, 40, 100)
    lnH_theory = wb.rho_ * np.log(t_theory) - wb.rho_ * np.log(wb.lambda_)
    ax.plot(np.log(t_theory), lnH_theory, '--', alpha=0.7)

ax.set_title('Проверка адекватности Вейбулла', fontsize=12)
ax.set_xlabel('ln(Время)')
ax.set_ylabel('ln(Кумулятивный риск)')
ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
from lifelines import KaplanMeierFitter

kmf_control = KaplanMeierFitter().fit(df_control['time'],
                                      df_control['event'],
                                      label='KM: Старый')
kmf_drug = KaplanMeierFitter().fit(df_drug['time'],
                                  df_drug['event'],
                                  label='KM: Новый')

fig, ax = plt.subplots(figsize=(10, 6))

# Непараметрические оценки
kmf_control.plot_survival_function(ax=ax, lw=1.5, alpha=0.7)
kmf_drug.plot_survival_function(ax=ax, lw=1.5, alpha=0.7)

# Параметрические оценки Вейбулла
wb_control.survival_function_.plot(ax=ax, lw=2.5, linestyle='--',
                                   label='Weibull: Старый')
wb_drug.survival_function_.plot(ax=ax, lw=2.5, linestyle='--',
                                label='Weibull: Новый')

ax.set_title('Функции выживаемости: Каплан-Мейер vs Вейбулл', fontsize=14)
ax.set_xlabel('Время (месяцы)')
ax.set_ylabel('Вероятность отсутствия рестеноза')
ax.legend()
ax.grid(alpha=0.3)
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
qq_plot(wb_control, ax=axes[0])
axes[0].set_title('Q-Q plot: Контроль')
qq_plot(wb_drug, ax=axes[1])
axes[1].set_title('Q-Q plot: Новый стент')
plt.tight_layout()
plt.show()
```

Что мы видим у контрольной группы?
$$\rho = 1.5 > 1,
$$
риск рестеноза растёт со временем. Это логично: через 6 месяцев покрытие заканчивается, начинается пролиферация эндотелия.

Медиана ≈ 15.2 месяца.

Что дало новое покрытие?

k
=
1.3
k=1.3 → риск растёт медленнее.

λ
=
24.0
λ=24.0 против 18.0 → «характерное время жизни» увеличено на 33%.

Медиана ≈ 19.8 месяца → выигрыш ~4.5 месяца до рестеноза у половины пациентов.

Ключевое для врача-кибернетика:

Параметр формы
k
k говорит о динамике процесса старения/износа.
Если
k
<
1
k<1 — риск снижается (адаптация, «хирургия удалила проблему»).
Если
k
>
1
k>1 — риск растет (деградация, накопление повреждений).
В нашем случае оба
k
>
1
k>1, но новое покрытие замедляет скорость нарастания риска.

```python
import numpy as np
import pandas as pd
from lifelines import CoxPHFitter
from lifelines.datasets import load_rossi

rossi = load_rossi()

results = {}
for p in np.linspace(0.001, 0.2, 40):
    cph = CoxPHFitter(l1_ratio=1., penalizer=p).fit(rossi, "week", "arrest")
    results[p] = cph.params_

pd.DataFrame(results).T.plot()
```

```python
def l_one_half_cox(lambda_, df, T, E):
    EPSILON = 0.00001

    weights = lambda_ * np.ones(df.shape[1]-2)
    cph = CoxPHFitter(l1_ratio=1.0, penalizer=weights)
    cph.fit(rossi, "week", "arrest")
    max_iter = 20
    i = 1

    while i < max_iter:
        weights = lambda_ / (np.sqrt((cph.params_.abs()).values) + EPSILON)
        cph = CoxPHFitter(l1_ratio=1.0, penalizer=weights)
        cph.fit(rossi, "week", "arrest")
        i += 1

    return cph.params_
```
